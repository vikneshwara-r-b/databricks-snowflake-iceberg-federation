```python
import requests
import json
import os
```

```python
dbutils.widgets.text("f1_season_year", "2024")
dbutils.widgets.text("f1_season_round", "1")
dbutils.widgets.text("f1_api_export_filename_prefix", "race_results_export")
dbutils.widgets.text("volume_root_path", "")
dbutils.widgets.text("volume_subdir_path", "race_results_dumps")
dbutils.widgets.text("f1_api_base_url", "https://f1api.dev/api")

# F1 Race API Base URL
f1_api_base_url = dbutils.widgets.get("f1_api_base_url")
print(f"F1 Base API URL: {f1_api_base_url}")

f1_season_year = dbutils.widgets.get("f1_season_year")
print(f"F1 Season Year: {f1_season_year}")

f1_season_round = dbutils.widgets.get("f1_season_round")
print(f"F1 Season Round: {f1_season_round}")

api_export_filename_prefix = dbutils.widgets.get("f1_api_export_filename_prefix")
print(f"API Export Filename Prefix: {api_export_filename_prefix}")

volume_root_path = dbutils.widgets.get("volume_root_path")
print(f"Volume Root Path: {volume_root_path}")

volume_subdir_path = dbutils.widgets.get("volume_subdir_path")
print(f"Volume Subdir Path: {volume_subdir_path}")
```

```python
output_file_name = "{0}_{1}_{2}.json".format(api_export_filename_prefix, f1_season_year, f1_season_round)
volume_directory = os.path.join(volume_root_path, volume_subdir_path)
output_file_path = os.path.join(volume_directory, output_file_name)

# Create the directory safely
os.makedirs(volume_directory, exist_ok=True)
```

```python
url = os.path.join(f1_api_base_url, f1_season_year, f1_season_round, "race")
print(f"F1 Race API URL: {url}")
raw_payload = requests.get(url)

output_dict = raw_payload.json()
output_dict
```

```python
# Write the output_dict to the volume as JSON
with open(output_file_path, 'w') as f:
    json.dump(output_dict, f, indent=2)

print(f"Data saved to {output_file_path}")
```
